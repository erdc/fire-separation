```python
def get_poly_sqr_from_bounds(active_zone_bounds):
    
    active_zone_ll = Polygon([[active_zone_bounds[0][0],active_zone_bounds[1][0]],
                              [active_zone_bounds[0][1],active_zone_bounds[1][0]],
                              [active_zone_bounds[0][1],active_zone_bounds[1][1]],
                              [active_zone_bounds[0][0],active_zone_bounds[1][1]],
                              [active_zone_bounds[0][0],active_zone_bounds[1][0]]])

    return active_zone_ll

def get_active_area_zones_form_bounds(active_zone_bounds):
    active_zone_im = np.array(((21600-active_zone_bounds[1][1]*240, 21600-active_zone_bounds[1][0]*240),
                           (active_zone_bounds[0][0]*240, active_zone_bounds[0][1]*240)))#np.array(active_zone_bounds)*240

    active_zone_im = active_zone_im.astype(int)

    return active_zone_im

def get_land_use_from_pixel_bbox(active_zone_bounds):
    # Get_converted_zones
    active_zone_im = get_active_area_zones_form_bounds(active_zone_bounds)
    
    # Open Land Use File
    im = np.array(Image.open('../data/gm_lc_v3/gm_lc_v3_1_2.tif'))
    print('Image size: {}, lat scale factor: {}, lng scale factor: {}'.format(im.shape,im.shape[0]/90, im.shape[1]/180))
    
    im = im[active_zone_im[0][0]: active_zone_im[0][1], active_zone_im[1][0]: active_zone_im[1][1]]
    
    # Open Ukraine Border Filem
    #shapefile = gpd.read_file("../data/ukraine_border/nv937bq8361.shp").iloc[0].geoms[-1]   # The last polygon has the largest area. This is used as 'Ukraine'
    
    return im, im.shape, active_zone_im
```

## Open Street Maps Land Use Data

```python
def haversine(lon1, lat1, lon2, lat2):
    """
    Calculate the great circle distance in kilometers between two points 
    on the earth (specified in decimal degrees)
    """
    # convert decimal degrees to radians 
    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])

    # haversine formula 
    dlon = lon2 - lon1 
    dlat = lat2 - lat1 
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * asin(sqrt(a)) 
    r = 3956 #6371 # Radius of earth in kilometers. Use 3956 for miles. Determines return value units.
    return c * r
```

```python
geometry_type = 'grid' # grid or polygons

tags = {'landuse': ['forest'], 'natural': ['wood']} #

active_zone_ll = get_poly_sqr_from_bounds(active_zone_bounds)

geometries = ox.geometries_from_bbox(active_zone_bounds[1][0], active_zone_bounds[1][1], 
                                     active_zone_bounds[0][0], active_zone_bounds[0][1], tags=tags)

geometries.reset_index(inplace = True)

if geometry_type == 'grid':
    
    div_factor = 0.5 #0.6 #0.9
    active_zone_bounds =  ((-122.00, -121.75),(37.00, 37.25))
    points = []
    x_points = int(100*(active_zone_bounds[0][1] - active_zone_bounds[0][0])/div_factor) 
    xy_factor = haversine(active_zone_bounds[0][0], active_zone_bounds[1][0], active_zone_bounds[0][0], active_zone_bounds[1][1])/haversine(active_zone_bounds[0][0], active_zone_bounds[1][0], active_zone_bounds[0][1], active_zone_bounds[1][0]) 

    for x in range(x_points + 1):
        for y in range(int(x_points*xy_factor) + 1):
            point = Point(active_zone_bounds[0][0] + x/x_points*(active_zone_bounds[0][1] - active_zone_bounds[0][0]),
                          active_zone_bounds[1][0] + y/int(x_points*xy_factor)*(active_zone_bounds[0][1] - active_zone_bounds[0][0]))

            if np.sum(geometries.apply(lambda row:row['geometry'].intersects(point), axis=1)) > 0:
                points.append(point)

    geometries = pd.DataFrame(points, columns=['geometry'])

len(geometries), geometries.head(2), geometries.tail(2)
```

```python
# USGS Elevation Point Query Service
url = r'https://epqs.nationalmap.gov/v1/json?'

# coordinates with known elevation 
lat = [centroid.y for centroid in centroids]
lon = [centroid.x for centroid in centroids]

# create data frame
elevation_df = pd.DataFrame({
    'lat': lat,
    'lon': lon
})

def elevation_function(df, lat_column, lon_column):
    """Query service using lat, lon. add the elevation values as a new column."""
    elevations = []
    counter = 0
    for lat, lon in zip(df[lat_column], df[lon_column]):
        if counter % 100 == 0:
            print(counter)
        # define rest query params
        params = {
            'output': 'json',
            'x': lon,
            'y': lat,
            'units': 'Meters'}
        
        # format query string and return query value
        
        #elevations.append(result.json()['USGS_Elevation_Point_Query_Service']['Elevation_Query']['Elevation'])
        #new 2023:
        try:
            result = requests.get((url + urllib.parse.urlencode(params)))
            elevations.append(result.json()['value'])
        except Exception as e:
            print(f"Error for (lat, lon) ({lat}, {lon}):",e)
            elevations.append(-500)
        counter += 1

    elevation_df['elev_meters'] = elevations

elevation_function(elevation_df, 'lat', 'lon')
elevation_df.head()

#### KS TODO: Convert Meter to Feet!!!
```

```python
elev_to_max_dist = {40: {25: 1.73, 50: 1.42, 100: 1.11, 200: 0.79},# wind speed: {effective tree height: max distance}
                    30: {25: 1.29, 50: 1.07, 100: 0.83, 200: 0.60},
                    20: {25: 0.87, 50: 0.72, 100: 0.57, 200: 0.42},
                    10: {25: 0.44, 50: 0.36, 100: 0.29, 200: 0.20}}

def max_spotting_distance(wind_speed, difference_in_elevation):
    slope, intercept, r_value, p_value, std_err = stats.linregress(list(elev_to_max_dist[wind_speed].keys()),
                                                                   list(elev_to_max_dist[wind_speed].values()))

    difference_in_elevation = 3.28084*difference_in_elevation # convert from meters to feet
    
    return slope*(difference_in_elevation) + intercept # returns results in miles
```

```python
G = nx.Graph()
#cutoff_distance = 1.0 # miles
wind_speed = 20

centroids = []

for tree_num1, tree_geom in enumerate(geometries['geometry']):
    if tree_num1 % 100 == 0:
        print(tree_num1)
        
    if type(geometries['geometry'][tree_num1]) == Point:
        lon1 = geometries['geometry'][tree_num1].coords.xy[0][0]
        lat1 = geometries['geometry'][tree_num1].coords.xy[1][0]
        centroids.append(geometries['geometry'][tree_num1])
    else:
        lon1 = geometries['geometry'][tree_num1].centroid.coords.xy[0][0]
        lat1 = geometries['geometry'][tree_num1].centroid.coords.xy[1][0]
        centroids.append(geometries['geometry'][tree_num1].centroid)
    
    G.add_node(tree_num1, pos=(lon1, lat1))
    
    for tree_num2 in range(0,tree_num1):
        if type(geometries['geometry'][tree_num2]) == Point:
            lon2 = geometries['geometry'][tree_num2].coords.xy[0][0]
            lat2 = geometries['geometry'][tree_num2].coords.xy[1][0]
        else:
            lon2 = geometries['geometry'][tree_num2].centroid.coords.xy[0][0]
            lat2 = geometries['geometry'][tree_num2].centroid.coords.xy[1][0]

        ## Cutoff for Edge
        distance = haversine(lon1, lat1, lon2, lat2)

        difference_in_elevation = abs(elevation_df.loc[tree_num1, 'elev_meters'] - elevation_df.loc[tree_num2, 'elev_meters'])
        cutoff_distance = max_spotting_distance(wind_speed, difference_in_elevation)
        # if tree_num1 % 100 == 0 and tree_num2 % 100 == 0:
        #     print(f"\telev. diff: {3.28084*difference_in_elevation}\n\tCutoff: {cutoff_distance}")

        ## Could change to edge of polygons instead of centroids -> higher connectivity
        if distance < cutoff_distance:
            #print('included, distance: {}, cutoff: {}'.format(distance, cutoff_distance))
            G.add_edge(tree_num1, tree_num2)
```
