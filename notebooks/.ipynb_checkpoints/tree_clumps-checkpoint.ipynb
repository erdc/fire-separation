```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import geopandas as gpd
from shapely import affinity, Polygon, Point
import folium
from folium import plugins
from folium.plugins import HeatMap
import osmnx as ox

Image.MAX_IMAGE_PIXELS = None
```

## Define Active Area

```python
active_zone_bounds = ((37.85, 38.2),(48.35, 48.7))

active_zone_bounds
```

## Land Use Data

Files: 
* 1_1 = North West (North America)
* 1_2 = North East (Europe & North Asia

```python
active_zone_im[0][0], active_zone_im[0][1], active_zone_im[1][0], active_zone_im[1][1]
```

```python
def get_poly_sqr_from_bounds(active_zone_bounds):
    
    active_zone_ll = Polygon([[active_zone_bounds[0][0],active_zone_bounds[1][0]],
                              [active_zone_bounds[0][1],active_zone_bounds[1][0]],
                              [active_zone_bounds[0][1],active_zone_bounds[1][1]],
                              [active_zone_bounds[0][0],active_zone_bounds[1][1]],
                              [active_zone_bounds[0][0],active_zone_bounds[1][0]]])

    return active_zone_ll

def get_active_area_zones_form_bounds(active_zone_bounds):
    active_zone_im = np.array(((21600-active_zone_bounds[1][1]*240, 21600-active_zone_bounds[1][0]*240),
                           (active_zone_bounds[0][0]*240, active_zone_bounds[0][1]*240)))#np.array(active_zone_bounds)*240

    active_zone_im = active_zone_im.astype(int)

    return active_zone_im

def get_land_use_from_pixel_bbox(active_zone_bounds):
    # Get_converted_zones
    active_zone_im = get_active_area_zones_form_bounds(active_zone_bounds)
    
    # Open Land Use File
    im = np.array(Image.open('../data/gm_lc_v3/gm_lc_v3_1_2.tif'))
    print('Image size: {}, lat scale factor: {}, lng scale factor: {}'.format(im.shape,im.shape[0]/90, im.shape[1]/180))
    
    im = im[active_zone_im[0][0]: active_zone_im[0][1], active_zone_im[1][0]: active_zone_im[1][1]]
    
    # Open Ukraine Border Filem
    #shapefile = gpd.read_file("../data/ukraine_border/nv937bq8361.shp").iloc[0].geoms[-1]   # The last polygon has the largest area. This is used as 'Ukraine'
    
    return im, im.shape, active_zone_im


# Get and Plot Land Use Data
ukraine_land_use, im_shape, active_zone_im = get_land_use_from_pixel_bbox(active_zone_bounds)

plt.imshow(ukraine_land_use)
```

```python
def get_heatmap_from_array_for_folium(ukraine_land_use, land_use_code):
    heatmap = []
    intensities = []
    
    for x in range(len(ukraine_land_use)):
        for y in range(len(ukraine_land_use[0])):
            intensities.append(float(ukraine_land_use[x][y]))
            if ukraine_land_use[x][y] == land_use_code:   # 5 = Trees
                heatmap.append([active_zone_ll.bounds[3] - (x/240) , (y)/240 + active_zone_ll.bounds[0], float(ukraine_land_use[x][y])/13])

    print('Unique landuse codes: {}, selected: {}'.format(np.unique(intensities), land_use_code))

    return heatmap
```

```python
lon, lat = active_zone_ll.centroid.coords[0]
zoom_start_val = 9

# Basemap
m = folium.Map([lat, lon], tiles='OpenStreetMap', zoom_start=zoom_start_val)#, tiles='OpenStreetMap')

# Bounding box around active zone = Red
active_zone_ll = get_poly_sqr_from_bounds(active_zone_bounds)
sim_geo = gpd.GeoSeries(active_zone_ll).simplify(tolerance=0.001)
geo_j = sim_geo.to_json()
geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"fill":False,"color": "red"})#orange"})
geo_j.add_to(m)

# Heatmap of selected land use code
heatmap = get_heatmap_from_array_for_folium(ukraine_land_use, 5)

HeatMap(heatmap,#gradient={0.1: 'blue', 0.3: 'lime', 0.5: 'yellow', 0.7: 'orange', 1: 'red'}, 
        use_local_extrema=False,
        radius=7#,zmin=1.0,zmax=15.0
       ).add_to(folium.FeatureGroup(name='Heat Map').add_to(m))
folium.LayerControl().add_to(m)

m
```

## Open Street Maps Land Use Data

```python
tags = {'landuse': ['forest'], 'natural': ['wood']} #

geometries = ox.geometries_from_bbox(active_zone_bounds[1][0], active_zone_bounds[1][1], 
                                     active_zone_bounds[0][0], active_zone_bounds[0][1], tags=tags)

geometries.reset_index(inplace = True)
```

```python
lon, lat = active_zone_ll.centroid.coords[0]
zoom_start_val = 11
min_area = 1e-6
polygons = []
centroids = []


# Basemap
m = folium.Map([lat, lon], tiles='OpenStreetMap', zoom_start=zoom_start_val)#, tiles='OpenStreetMap')

# Bounding box around active zone = Red
active_zone_ll = get_poly_sqr_from_bounds(active_zone_bounds)
sim_geo = gpd.GeoSeries(active_zone_ll).simplify(tolerance=0.001)
geo_j = sim_geo.to_json()
geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"fill":False,"color": "red"})#orange"})
geo_j.add_to(m)

# Plot OSM Results
for tree_geom in geometries['geometry']:
    if tree_geom.area > min_area:
        # Plot Polygon
        sim_geo = gpd.GeoSeries(tree_geom)#.simplify(tolerance=0.001)
        geo_j = sim_geo.to_json()
        geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"color": "green", "width": 2})#orange"})
        geo_j.add_to(m)
        polygons.append(tree_geom)

        # Plot Centroids
        if tree_geom.geom_type == 'Polygon':
            centroid = tree_geom.centroid
        elif tree_geom.geom_type == 'Point':
            centroid = tree_geom

        folium.CircleMarker(location=[centroid.y, centroid.x],
                            radius=0.2,
                            color = 'blue').add_to(m)#weight=5, color = 'green').add_to(m)

        centroids.append(centroid)
m
```

```python
lon, lat = active_zone_ll.centroid.coords[0]
zoom_start_val = 9

# Basemap
m = folium.Map([lat, lon], tiles='OpenStreetMap', zoom_start=zoom_start_val)#, tiles='OpenStreetMap')

# Bounding box around active zone = Red
active_zone_ll = get_poly_sqr_from_bounds(active_zone_bounds)
sim_geo = gpd.GeoSeries(active_zone_ll).simplify(tolerance=0.001)
geo_j = sim_geo.to_json()
geo_j = folium.GeoJson(data=geo_j, style_function=lambda x: {"fill":False,"color": "red"})#orange"})
geo_j.add_to(m)

# Plot OSM Results
for tree_geom in geometries['geometry']:

    if tree_geom.area > min_area:
        if tree_geom.geom_type == 'Polygon':
            centroid = tree_geom.centroid
        elif tree_geom.geom_type == 'Point':
            centroid = tree_geom

        folium.CircleMarker(location=[centroid.y, centroid.x],
                            radius=0.1,
                            color = 'green').add_to(m)#weight=5, color = 'green').add_to(m)

m
```

```python
trees_dict = {'geometry': polygons, 'centroids': centroids} 
    
trees_df = pd.DataFrame(trees_dict)
trees_df.to_csv('../data/results/identified_trees.csv')
```
